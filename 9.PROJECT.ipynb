```python
# ============================================================
#        DATA ANALYSIS & VISUALIZATION PROJECT
# ============================================================
# Technologies:
# 1. Python
# 2. NumPy
# 3. Pandas
# 4. Matplotlib
# 5. Seaborn
# 6. CSV File Handling
# 7. Data Analysis
# 8. Data Visualization
# 9. Data Validation
# ============================================================

import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns


# ============================================================
# GLOBAL VARIABLES
# ============================================================

df = None
current_plot = None

# CSV file path
DATA_FILE = "data/sales_data.csv"


# ============================================================
# MAIN MENU
# ============================================================

def display_menu():

    print("\n========== Data Analysis & Visualization Program ==========")

    print("Please select an option:")
    print("1. Load Dataset")
    print("2. Explore Data")
    print("3. Perform DataFrame Operations")
    print("4. Handle Missing Data")
    print("5. Generate Descriptive Statistics")
    print("6. Data Visualization")
    print("7. Save Visualization")
    print("8. Exit")

    print("===========================================================")


# ============================================================
# CHECK DATASET
# ============================================================

def check_dataset():

    if df is None:

        print("\nPlease load the dataset first!")

        return False

    return True


# ============================================================
# LOAD DATASET
# ============================================================

def load_dataset():

    global df

    print("\n== Load Dataset ==")

    path = input(
        "Enter the path of the dataset (CSV file): "
    ).strip()

    # If user directly presses Enter
    if path == "":
        path = DATA_FILE

    try:

        # Check file exists
        if not os.path.exists(path):

            print("Error: File not found!")

            return

        # Read CSV
        loaded_df = pd.read_csv(path)

        # Required columns
        required_columns = [
            "SalesID",
            "Product",
            "Region",
            "Sales",
            "Year"
        ]

        # Check required columns
        missing_columns = [
            column
            for column in required_columns
            if column not in loaded_df.columns
        ]

        if missing_columns:

            print("Error: Required columns are missing:")

            print(missing_columns)

            return

        # Convert Sales to numeric
        loaded_df["Sales"] = pd.to_numeric(
            loaded_df["Sales"],
            errors="coerce"
        )

        # Convert Year to numeric
        loaded_df["Year"] = pd.to_numeric(
            loaded_df["Year"],
            errors="coerce"
        )

        # Remove invalid rows
        loaded_df.dropna(
            subset=[
                "SalesID",
                "Product",
                "Region",
                "Sales",
                "Year"
            ],
            inplace=True
        )

        # Convert ID and Year to integer
        loaded_df["SalesID"] = loaded_df["SalesID"].astype(int)

        loaded_df["Year"] = loaded_df["Year"].astype(int)

        # Check negative sales
        if (loaded_df["Sales"] < 0).any():

            print(
                "Error: Sales cannot contain negative values."
            )

            return

        # Store data
        df = loaded_df

        print("Dataset loaded successfully!")

        print("Total records:", len(df))

        print("Total columns:", len(df.columns))

    except Exception as e:

        print("Error while loading dataset:", e)


# ============================================================
# EXPLORE DATA
# ============================================================

def explore_data():

    if not check_dataset():

        return

    while True:

        print("\n== Explore Data ==")

        print("1. Display the first 5 rows")
        print("2. Display the last 5 rows")
        print("3. Display column names")
        print("4. Display data types")
        print("5. Display basic info")
        print("6. Display dataset shape")
        print("7. Display unique values")
        print("8. Back to Main Menu")

        choice = input("Enter your choice: ").strip()

        # ----------------------------------------------------
        # FIRST 5 ROWS
        # ----------------------------------------------------

        if choice == "1":

            print("\n")

            print(df.head())

        # ----------------------------------------------------
        # LAST 5 ROWS
        # ----------------------------------------------------

        elif choice == "2":

            print("\n")

            print(df.tail())

        # ----------------------------------------------------
        # COLUMN NAMES
        # ----------------------------------------------------

        elif choice == "3":

            print("\nColumn Names:")

            for column in df.columns:

                print("-", column)

        # ----------------------------------------------------
        # DATA TYPES
        # ----------------------------------------------------

        elif choice == "4":

            print("\nData Types:")

            print(df.dtypes)

        # ----------------------------------------------------
        # BASIC INFO
        # ----------------------------------------------------

        elif choice == "5":

            print("\nBasic Information:")

            df.info()

        # ----------------------------------------------------
        # SHAPE
        # ----------------------------------------------------

        elif choice == "6":

            print("\nDataset Shape:")

            print("Rows:", df.shape[0])

            print("Columns:", df.shape[1])

        # ----------------------------------------------------
        # UNIQUE VALUES
        # ----------------------------------------------------

        elif choice == "7":

            print("\nUnique Values:")

            for column in df.columns:

                print("\n" + column + ":")

                print(df[column].unique())

        # ----------------------------------------------------
        # BACK
        # ----------------------------------------------------

        elif choice == "8":

            break

        else:

            print("Invalid choice! Please try again.")


# ============================================================
# DATAFRAME OPERATIONS
# ============================================================

def dataframe_operations():

    if not check_dataset():

        return

    while True:

        print("\n== Perform DataFrame Operations ==")

        print("1. Sort by Sales")
        print("2. Sort by Year")
        print("3. Filter Sales greater than a value")
        print("4. Filter by Region")
        print("5. Select specific columns")
        print("6. Add a new calculated column")
        print("7. Group by Region")
        print("8. Group by Product")
        print("9. Back to Main Menu")

        choice = input("Enter your choice: ").strip()

        # ----------------------------------------------------
        # SORT SALES
        # ----------------------------------------------------

        if choice == "1":

            result = df.sort_values(
                by="Sales",
                ascending=False
            )

            print("\nData sorted by Sales:")

            print(result)

        # ----------------------------------------------------
        # SORT YEAR
        # ----------------------------------------------------

        elif choice == "2":

            result = df.sort_values(
                by="Year"
            )

            print("\nData sorted by Year:")

            print(result)

        # ----------------------------------------------------
        # FILTER SALES
        # ----------------------------------------------------

        elif choice == "3":

            try:

                value = float(
                    input(
                        "Enter minimum Sales value: "
                    )
                )

                result = df[
                    df["Sales"] > value
                ]

                print("\nFiltered Data:")

                print(result)

            except ValueError:

                print(
                    "Please enter a valid number."
                )

        # ----------------------------------------------------
        # FILTER REGION
        # ----------------------------------------------------

        elif choice == "4":

            region = input(
                "Enter Region: "
            ).strip()

            result = df[
                df["Region"]
                .astype(str)
                .str.lower()
                ==
                region.lower()
            ]

            print("\nFiltered Data:")

            print(result)

        # ----------------------------------------------------
        # SELECT COLUMNS
        # ----------------------------------------------------

        elif choice == "5":

            print("\nAvailable columns:")

            print(", ".join(df.columns))

            columns = input(
                "Enter column names separated by comma: "
            ).split(",")

            columns = [
                column.strip()
                for column in columns
            ]

            if all(
                column in df.columns
                for column in columns
            ):

                print(df[columns])

            else:

                print("Invalid column name!")

        # ----------------------------------------------------
        # NEW CALCULATED COLUMN
        # ----------------------------------------------------

        elif choice == "6":

            df["Sales_with_10_percent"] = (
                df["Sales"] * 1.10
            )

            print("\nNew column added:")

            print(
                df[
                    [
                        "Sales",
                        "Sales_with_10_percent"
                    ]
                ].head()
            )

        # ----------------------------------------------------
        # GROUP BY REGION
        # ----------------------------------------------------

        elif choice == "7":

            result = df.groupby(
                "Region"
            )["Sales"].agg(
                ["sum", "mean", "count"]
            )

            print("\nSales by Region:")

            print(result)

        # ----------------------------------------------------
        # GROUP BY PRODUCT
        # ----------------------------------------------------

        elif choice == "8":

            result = df.groupby(
                "Product"
            )["Sales"].agg(
                ["sum", "mean", "count"]
            )

            print("\nSales by Product:")

            print(result)

        # ----------------------------------------------------
        # BACK
        # ----------------------------------------------------

        elif choice == "9":

            break

        else:

            print("Invalid choice! Please try again.")


# ============================================================
# HANDLE MISSING DATA
# ============================================================

def handle_missing_data():

    global df

    if not check_dataset():

        return

    while True:

        print("\n== Handle Missing Data ==")

        print("1. Display rows with missing values")
        print("2. Fill missing values with mean")
        print("3. Drop rows with missing values")
        print("4. Replace missing values with a specific value")
        print("5. Display missing value count")
        print("6. Back to Main Menu")

        choice = input("Enter your choice: ").strip()

        # ----------------------------------------------------
        # DISPLAY MISSING ROWS
        # ----------------------------------------------------

        if choice == "1":

            missing_rows = df[
                df.isnull().any(axis=1)
            ]

            if missing_rows.empty:

                print(
                    "\nNo missing values found in the dataset!"
                )

            else:

                print(
                    "\nRows with missing values:"
                )

                print(missing_rows)

        # ----------------------------------------------------
        # FILL WITH MEAN
        # ----------------------------------------------------

        elif choice == "2":

            numeric_columns = df.select_dtypes(
                include=np.number
            ).columns

            for column in numeric_columns:

                df[column] = df[column].fillna(
                    df[column].mean()
                )

            print(
                "Missing numeric values filled with mean successfully!"
            )

        # ----------------------------------------------------
        # DROP MISSING ROWS
        # ----------------------------------------------------

        elif choice == "3":

            before = len(df)

            df.dropna(
                inplace=True
            )

            after = len(df)

            print(
                f"Dropped {before - after} rows."
            )

            print(
                "Missing rows removed successfully!"
            )

        # ----------------------------------------------------
        # REPLACE MISSING
        # ----------------------------------------------------

        elif choice == "4":

            value = input(
                "Enter value to replace missing data: "
            ).strip()

            if value == "":

                print(
                    "Value cannot be empty."
                )

            else:

                df.fillna(
                    value,
                    inplace=True
                )

                print(
                    "Missing values replaced successfully!"
                )

        # ----------------------------------------------------
        # MISSING COUNT
        # ----------------------------------------------------

        elif choice == "5":

            print("\nMissing Values Count:")

            print(
                df.isnull().sum()
            )

        # ----------------------------------------------------
        # BACK
        # ----------------------------------------------------

        elif choice == "6":

            break

        else:

            print(
                "Invalid choice! Please try again."
            )


# ============================================================
# DESCRIPTIVE STATISTICS
# ============================================================

def descriptive_statistics():

    if not check_dataset():

        return

    print(
        "\n== Generate Descriptive Statistics =="
    )

    numeric_df = df.select_dtypes(
        include=np.number
    )

    print(
        "\nDescriptive Statistics:"
    )

    print(
        numeric_df.describe()
    )

    # Sales analysis
    if "Sales" in df.columns:

        sales = df["Sales"].to_numpy()

        print("\nSales Analysis:")

        print(
            "Total Sales   :",
            np.sum(sales)
        )

        print(
            "Average Sales :",
            np.mean(sales)
        )

        print(
            "Maximum Sales :",
            np.max(sales)
        )

        print(
            "Minimum Sales :",
            np.min(sales)
        )

        print(
            "Standard Dev. :",
            np.std(sales)
        )

        print("\nSales by Region:")

        print(
            df.groupby(
                "Region"
            )["Sales"].sum()
        )

        print("\nSales by Product:")

        print(
            df.groupby(
                "Product"
            )["Sales"].sum()
        )


# ============================================================
# DATA VISUALIZATION MENU
# ============================================================

def data_visualization():

    if not check_dataset():

        return

    while True:

        print("\n== Data Visualization ==")

        print("1. Bar Plot")
        print("2. Line Plot")
        print("3. Scatter Plot")
        print("4. Pie Chart")
        print("5. Histogram")
        print("6. Stack Plot")
        print("7. Back to Main Menu")

        choice = input(
            "Enter your choice: "
        ).strip()

        try:

            if choice == "1":

                bar_plot()

            elif choice == "2":

                line_plot()

            elif choice == "3":

                scatter_plot()

            elif choice == "4":

                pie_chart()

            elif choice == "5":

                histogram()

            elif choice == "6":

                stack_plot()

            elif choice == "7":

                break

            else:

                print(
                    "Invalid choice! Please try again."
                )

        except Exception as e:

            print(
                "Unable to generate visualization:",
                e
            )


# ============================================================
# BAR PLOT
# ============================================================

def bar_plot():

    global current_plot

    print("\n== Bar Plot ==")

    grouped = df.groupby(
        "Product"
    )["Sales"].sum()

    plt.figure(
        figsize=(10, 6)
    )

    sns.barplot(
        x=grouped.index,
        y=grouped.values
    )

    plt.title(
        "Total Sales by Product"
    )

    plt.xlabel(
        "Product"
    )

    plt.ylabel(
        "Sales"
    )

    plt.tight_layout()

    current_plot = plt.gcf()

    print(
        "Generating bar plot..."
    )

    plt.show()

    print(
        "Bar plot displayed successfully!"
    )


# ============================================================
# LINE PLOT
# ============================================================

def line_plot():

    global current_plot

    print("\n== Line Plot ==")

    yearly_sales = df.groupby(
        "Year"
    )["Sales"].sum()

    plt.figure(
        figsize=(10, 6)
    )

    plt.plot(
        yearly_sales.index,
        yearly_sales.values,
        marker="o",
        linewidth=2
    )

    plt.title(
        "Sales Trend by Year"
    )

    plt.xlabel(
        "Year"
    )

    plt.ylabel(
        "Total Sales"
    )

    plt.grid(True)

    plt.tight_layout()

    current_plot = plt.gcf()

    print(
        "Generating line plot..."
    )

    plt.show()

    print(
        "Line plot displayed successfully!"
    )


# ============================================================
# SCATTER PLOT
# ============================================================

def scatter_plot():

    global current_plot

    print("\n== Scatter Plot ==")

    print(
        "Available numeric columns:"
    )

    print(
        df.select_dtypes(
            include=np.number
        ).columns.tolist()
    )

    x_column = input(
        "Enter x-axis column name: "
    ).strip()

    y_column = input(
        "Enter y-axis column name: "
    ).strip()

    # Check columns
    if (
        x_column not in df.columns
        or
        y_column not in df.columns
    ):

        print(
            "Invalid column name!"
        )

        return

    # Check numeric columns
    if not (
        pd.api.types.is_numeric_dtype(
            df[x_column]
        )
        and
        pd.api.types.is_numeric_dtype(
            df[y_column]
        )
    ):

        print(
            "Both selected columns must contain numeric data!"
        )

        return

    print(
        "Generating scatter plot..."
    )

    plt.figure(
        figsize=(10, 6)
    )

    sns.scatterplot(
        data=df,
        x=x_column,
        y=y_column,
        s=100
    )

    plt.title(
        f"Scatter Plot: {x_column} vs {y_column}"
    )

    plt.xlabel(
        x_column
    )

    plt.ylabel(
        y_column
    )

    plt.grid(True)

    plt.tight_layout()

    current_plot = plt.gcf()

    plt.show()

    print(
        "Scatter plot displayed successfully!"
    )


# ============================================================
# PIE CHART
# ============================================================

def pie_chart():

    global current_plot

    print("\n== Pie Chart ==")

    grouped = df.groupby(
        "Region"
    )["Sales"].sum()

    plt.figure(
        figsize=(8, 8)
    )

    plt.pie(
        grouped.values,
        labels=grouped.index,
        autopct="%1.1f%%",
        startangle=90
    )

    plt.title(
        "Sales Distribution by Region"
    )

    plt.tight_layout()

    current_plot = plt.gcf()

    plt.show()

    print(
        "Pie chart displayed successfully!"
    )


# ============================================================
# HISTOGRAM
# ============================================================

def histogram():

    global current_plot

    print("\n== Histogram ==")

    plt.figure(
        figsize=(10, 6)
    )

    plt.hist(
        df["Sales"],
        bins=8,
        edgecolor="black"
    )

    plt.title(
        "Sales Distribution"
    )

    plt.xlabel(
        "Sales"
    )

    plt.ylabel(
        "Frequency"
    )

    plt.grid(True)

    plt.tight_layout()

    current_plot = plt.gcf()

    plt.show()

    print(
        "Histogram displayed successfully!"
    )


# ============================================================
# STACK PLOT
# ============================================================

def stack_plot():

    global current_plot

    print("\n== Stack Plot ==")

    pivot = df.pivot_table(
        values="Sales",
        index="Year",
        columns="Region",
        aggfunc="sum",
        fill_value=0
    )

    plt.figure(
        figsize=(12, 7)
    )

    plt.stackplot(
        pivot.index,
        *[
            pivot[column].values
            for column in pivot.columns
        ],
        labels=pivot.columns
    )

    plt.title(
        "Sales by Region and Year"
    )

    plt.xlabel(
        "Year"
    )

    plt.ylabel(
        "Sales"
    )

    plt.legend(
        loc="upper left"
    )

    plt.tight_layout()

    current_plot = plt.gcf()

    plt.show()

    print(
        "Stack plot displayed successfully!"
    )


# ============================================================
# SAVE VISUALIZATION
# ============================================================

def save_visualization():

    global current_plot

    print("\n== Save Visualization ==")

    if current_plot is None:

        print(
            "No visualization available to save!"
        )

        print(
            "Please generate a visualization first."
        )

        return

    file_name = input(
        "Enter file name to save the plot "
        "(e.g., scatter_plot.png): "
    ).strip()

    if file_name == "":

        print(
            "File name cannot be empty!"
        )

        return

    # If extension is not provided
    if not os.path.splitext(file_name)[1]:

        file_name += ".png"

    try:

        current_plot.savefig(
            file_name,
            dpi=300,
            bbox_inches="tight"
        )

        print(
            f"Visualization saved as "
            f"{file_name} successfully!"
        )

    except Exception as e:

        print(
            "Error while saving visualization:",
            e
        )


# ============================================================
# MAIN PROGRAM
# ============================================================

def main():

    while True:

        display_menu()

        choice = input(
            "\nEnter your choice: "
        ).strip()

        # ----------------------------------------------------
        # LOAD DATASET
        # ----------------------------------------------------

        if choice == "1":

            load_dataset()

        # ----------------------------------------------------
        # EXPLORE DATA
        # ----------------------------------------------------

        elif choice == "2":

            explore_data()

        # ----------------------------------------------------
        # DATAFRAME OPERATIONS
        # ----------------------------------------------------

        elif choice == "3":

            dataframe_operations()

        # ----------------------------------------------------
        # MISSING DATA
        # ----------------------------------------------------

        elif choice == "4":

            handle_missing_data()

        # ----------------------------------------------------
        # STATISTICS
        # ----------------------------------------------------

        elif choice == "5":

            descriptive_statistics()

        # ----------------------------------------------------
        # VISUALIZATION
        # ----------------------------------------------------

        elif choice == "6":

            data_visualization()

        # ----------------------------------------------------
        # SAVE
        # ----------------------------------------------------

        elif choice == "7":

            save_visualization()

        # ----------------------------------------------------
        # EXIT
        # ----------------------------------------------------

        elif choice == "8":

            print(
                "\nExiting the program. Goodbye!"
            )

            break

        else:

            print(
                "\nInvalid choice! "
                "Please enter a number from 1 to 8."
            )


# ============================================================
# PROGRAM START
# ============================================================

if __name__ == "__main__":

    main()
```
